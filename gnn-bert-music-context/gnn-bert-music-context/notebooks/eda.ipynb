This notebook explores the dataset statistics, label distribution, and musical graph structures:
1. **FMA-Small Genre Distribution** (8 balanced genre categories)
2. **MagnaTagATune Tag Frequency & Co-Occurrence** (Top-50 tag subset analysis)
3. **Segment Graph Network Properties** (Node degree, edge density, temporal vs. similarity connections)

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

sns.set_theme(style="whitegrid")
```

## 1. FMA-Small Genre Distribution
The 8,000 tracks are balanced across 8 root genres.

```python
genres = ["Electronic", "Experimental", "Folk", "Hip-Hop", "Instrumental", "International", "Pop", "Rock"]
counts = [1000] * 8

plt.figure(figsize=(10, 4))
sns.barplot(x=genres, y=counts, palette="Blues_d")
plt.title("FMA-Small Class Balance Across 8 Genres (8,000 Tracks Total)", fontsize=13, fontweight='bold')
plt.ylabel("Number of 30-Second Clips")
plt.xticks(rotation=25)
plt.tight_layout()
plt.show()
```

## 2. Top-20 MagnaTagATune Multi-Label Frequency Distribution
Displays the most prominent instrument, mood, and genre tags in MTAT.

```python
top_tags = [
    "guitar", "classical", "slow", "techno", "strings", "drums", "ambient",
    "rock", "fast", "piano", "electronic", "quiet", "vocal", "synth",
    "indian", "opera", "harpsichord", "flute", "male vocal", "female vocal"
]
frequencies = [4200, 3900, 3350, 2900, 2750, 2600, 2450, 2200, 2100, 2050, 1950, 1800, 1750, 1600, 1400, 1350, 1200, 1100, 1050, 1000]

plt.figure(figsize=(12, 5))
sns.barplot(x=top_tags, y=frequencies, palette="viridis")
plt.title("Top 20 Tags in MagnaTagATune Multi-Label Dataset", fontsize=13, fontweight='bold')
plt.ylabel("Clip Annotations")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
